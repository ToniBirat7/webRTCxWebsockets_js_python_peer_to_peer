**Why can't we use `express.listen()` for `Socket.io` server?**

We can't use `express.listen()` for `Socket.io` server because `express.listen()` creates an HTTP server that is not capable of handling WebSocket connections. `Socket.io` requires a server that can handle both HTTP and WebSocket protocols, which is why we need to create a separate HTTP server using the `http` module and then pass that server to `Socket.io`.

So, instead of using `express.listen()`, we create an HTTP server using the `http` module and then pass that server to `Socket.io` like this:

```javascript
const express = require("express");
const http = require("http");
const socketIo = require("socket.io");

const app = express();

const server = http.createServer(app);
const io = socketIo(server);

server.listen(3000, () => {
  console.log("Server is running on port 3000");
});
```

## **Intro to `Socket.IO`**

`Socket.IO` is a powerful library that enables real-time, bidirectional communication between web clients and servers. It is built on top of WebSockets and provides additional features such as automatic reconnection, multiplexing, and event-based communication.

**Why Do We Need `Socket.IO`?**

In traditional web applications, communication between the client and server is typically done through HTTP requests. This model is request-response based, meaning the client has to initiate a request to receive data from the server.

This type of communication is not ideal for real-time applications where data needs to be pushed from the server to the client without the client having to request it. The communication is `Unidirectional`, meaning data flows in one direction at a time.

But, what if we need `Bidirectional` communication, where both the client and server can send and receive data independently and in real-time?

And, keeps the connection `Persistent`, allowing continuous data exchange without the overhead of establishing new connections for each interaction?

This is where `Socket.IO` comes in. It allows for real-time, bidirectional communication between clients and servers, making it ideal for applications such as chat apps, live notifications, online gaming, and collaborative tools.

It uses `WebSockets` as the underlying transport mechanism but falls back to other techniques like long polling if WebSockets are not supported by the client or server.

**What is `WebSockets`?**

`WebSockets` is a protocol that provides full-duplex communication channels over a single TCP connection. It allows for persistent connections between the client and server, enabling real-time data exchange without the need for repeated HTTP requests.

With `WebSockets`, both the client and server can send messages to each other independently, making it suitable for real-time applications.

**Key Features of `Socket.IO`:**

1. **Real-time Communication**: Enables instant data exchange between clients and servers.

2. **Event-based Architecture**: Allows developers to define custom events for communication.

3. **Automatic Reconnection**: Handles connection drops and automatically attempts to reconnect.

4. **Multiplexing**: Supports multiple namespaces and rooms for organizing communication channels.

5. **Cross-browser Compatibility**: Works across different browsers and platforms.

6. **Binary Support**: Can handle binary data such as images and files.

## **Integrating Socket.IO**

Socket.IO is composed of two parts:

**1. A server that integrates with (or mounts on) the `Node.js` HTTP server.**

**2. A client library that loads on the browser side `socket.io-client` package.**

### **Setting Up the Server**

First install the `socket.io` package via npm:

```bash
npm install socket.io
```

Next, create a simple server using `Express` and integrate `Socket.IO`:

```javascript
import express from "express";
import { createServer } from "node:http";
import { fileURLToPath } from "node:url";
import { dirname, join } from "node:path";
import { Server } from "socket.io";

const app = express();
const server = createServer(app);
const io = new Server(server);

const __dirname = dirname(fileURLToPath(import.meta.url));

app.get("/", (req, res) => {
  res.sendFile(join(__dirname, "index.html"));
});

io.on("connection", (socket) => {
  console.log("a user connected");
});

io.on("connection", (socket) => {
  console.log("a user connected");
  socket.on("disconnect", () => {
    console.log("user disconnected");
  });
});

server.listen(3000, () => {
  console.log("server running at http://localhost:3000");
});
```

In this code, we create an Express server and integrate `Socket.IO` with it. We also set up a route to serve an HTML file.

**Client-Side (Frontend)**

Create an `index.html` file in the same directory with the following content:

```html
<script src="/socket.io/socket.io.js"></script>
<script>
  const socket = io();
</script>
```

That’s all it takes to load the `socket.io-client`, which exposes an `io` global (and the endpoint GET `/socket.io/socket.io.js`), and then connect.

You can also use a CDN instead of the local files (e.g. `<script src="https://cdn.socket.io/4.8.1/socket.io.min.js"></script>`).

### **Types of `Events` in Socket.IO**

Socket.IO uses an event-based architecture, allowing developers to define custom events for communication between the client and server. There are two main types of events:

**Built-in Events**: These are predefined events provided by Socket.IO, such as:

- `io.on('connection', (socket: Socket) => {})`: Triggered when a client connects to the server.

- `socket.on('disconnect', () => {})`: Triggered when a client disconnects from the server.

- `socket.on('error', (error) => {})`: Triggered when an error occurs.

- `socket.on('connect_error', (error) => {})`: Triggered when a connection error occurs.

**Custom Events**: Developers can define their own events for specific communication needs. For example:

**Server-side:**

```javascript
io.on("connection", (socket) => {
  socket.on("chat message", (msg) => {
    console.log("message: " + msg);
  });
});
```

**Client-side:**

```html
<script>
  const socket = io();
  socket.emit("chat message", "Hello, World!");
</script>
```

## **Emitting and Listening to Events**

In Socket.IO, communication between the client and server is done through events. You can emit events to send data and listen for events to receive data.

**What is `Emitting`?**

`Emitting` is the process of sending an event along with optional data from one side (client or server) to the other. You can emit events using the `emit` method.

**What is `Listening`?**

`Listening` is the process of waiting for and handling events that are emitted by the other side. You can listen for events using the `on` method.

### **Example: Emitting and Listening to Events**

**Server-side:**

```javascript
io.on("connection", (socket) => {
  console.log("a user connected");
  socket.on("chat message", (msg) => {
    console.log("message: " + msg);
    // Emit the message to all connected clients
    io.emit("chat message", msg);
  });
});
```

**Client-side:**

```html
<script>
  const socket = io();
  socket.on("chat message", (msg) => {
    console.log("message: " + msg);
  });
</script>
```

In this example, when a client emits a `chat message` event with a message, the server listens for that event, logs the message, and then emits the same message to all connected clients. Each client listens for the `chat message` event and logs the received message.

### **`Socket` Object**

When a client connects to the server, a `Socket` object is created for that specific connection. This object represents the individual connection between the client and server and provides methods for emitting and listening to events.

The `Socket` object is passed as an argument to the connection event handler on the server side:

```javascript
io.on("connection", (socket) => {
  // You can use the socket object here
});
```

You can use the `Socket` object to:

- Emit events to the specific client.

- Listen for events from the specific client.

**Example: Using the `Socket` Object**

```javascript
io.on("connection", (socket) => {
  console.log("a user connected");
  socket.on("chat message", (msg) => {
    console.log("message: " + msg);
    // Emit the message to the specific client
    socket.emit("chat message", msg);
  });
});
```

In this example, when a client emits a `chat message` event, the server listens for that event and then emits the same message back to that specific client using the `socket.emit` method.

**Note**: The `Socket` object is unique for each client connection, allowing for individualized communication between the server and each connected client.

### **Properties of the `Socket` Object**

Some commonly used properties of the `Socket` object include:

- `id`: A unique identifier for the socket connection.

- `connected`: A boolean indicating whether the socket is currently connected.

- `disconnected`: A boolean indicating whether the socket is currently disconnected.

- `rooms`: An object representing the rooms that the socket is currently in.

- `handshake`: An object containing information about the connection handshake, such as headers and query parameters.

**Example: Accessing `Socket` Properties**

```javascript
io.on("connection", (socket) => {
  console.log("Socket ID:", socket.id);
  console.log("Is Connected:", socket.connected);
});
```

In this example, when a client connects, the server logs the unique socket ID and the connection status of the socket.

## **Namespaces and Rooms in Socket.IO**

In Socket.IO, `Namespaces` and `Rooms` are two important concepts that help organize and manage communication channels between clients and servers.

**Namespaces**

Namespaces are a way to create separate communication channels within a single Socket.IO server. Each namespace has its own event handlers and can be used to isolate different parts of an application.

To create a namespace, you can use the `of` method on the `io` object:

```javascript
const chatNamespace = io.of("/chat");
chatNamespace.on("connection", (socket) => {
  console.log("a user connected to the chat namespace");
});
```

Clients can connect to a specific namespace by specifying the namespace in the connection URL:

```javascript
const socket = io("/chat");
```

**Rooms**

Rooms are a way to group sockets within a namespace. Sockets can join or leave rooms, and messages can be broadcasted to all sockets in a specific room.

To join a room, you can use the `join` method on the `Socket` object:

```javascript
socket.join("room1");
```

To leave a room, you can use the `leave` method:

```javascript
socket.leave("room1");
```

To broadcast a message to all sockets in a room, you can use the `to` method on the `io` object:

```javascript
io.to("room1").emit("message", "Hello, room1!");
```

**Example: Using Namespaces and Rooms**

```javascript
const chatNamespace = io.of("/chat");
chatNamespace.on("connection", (socket) => {
  console.log("a user connected to the chat namespace");
  socket.join("room1");
  socket.on("message", (msg) => {
    io.to("room1").emit("message", msg);
  });
});
```

In this example, we create a `chat` namespace and listen for connections. When a client connects, they join `room1`. When a message is received, it is broadcasted to all clients in `room1`.

### **Crystal Clear Difference Between Namespaces and Rooms**

Lets take an analogy of an office building:

- **Namespaces**: Think of namespaces as different floors in the office building. Each floor (namespace) can have its own set of rooms and employees (sockets). Employees on one floor cannot directly communicate with employees on another floor unless they go through a common area (the main server). This allows for separation of concerns and organization within the building.

- **Rooms**: Now, within each floor (namespace), there are different rooms where employees can gather for meetings or discussions. These rooms (rooms) allow employees to communicate with each other without involving the entire floor. Employees can join or leave these rooms as needed, and messages can be sent to all employees in a specific room without disturbing others on the same floor.

In summary, namespaces are like floors in a building that separate different areas of communication, while rooms are like meeting rooms within those floors that allow for more focused interactions among specific groups of employees.
