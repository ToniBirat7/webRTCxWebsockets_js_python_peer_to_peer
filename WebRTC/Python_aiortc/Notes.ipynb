## **`aiortc` Library for WebRTC in Python**

[Offical_Repo](https://github.com/aiortc/aiortc/tree/main)

`aiortc` is a library for Web Real-Time Communication (WebRTC) and Object Real-Time Communications (ORTC) in Python. It provides an implementation of the WebRTC and ORTC protocols, allowing developers to build real-time communication applications such as video conferencing, voice calls, and data sharing.

[Documentation_Link](https://aiortc.readthedocs.io/en/latest/api.html#module-aiortc)

### **Event's in `aiortc`**

**`@pc.on("track")`**

The `@pc.on("track")` decorator is used to register an event handler for the "track" event on a `RTCPeerConnection` object. This event is triggered when a new media track (audio or video) is added to the connection. The decorated function will be called with the track as an argument, allowing you to handle the incoming media stream.

```python
from aiortc import RTCPeerConnection, MediaStreamTrack

pc = RTCPeerConnection()

@pc.on("track")
async def on_track(track: MediaStreamTrack):
    print(f"📹 Track received: {track.kind}")
    if track.kind == "video":
        while True:
            frame = await track.recv()
            # do something with frame
```

**`@pc.on("datachannel")`**

The `@pc.on("datachannel")` decorator is used to register an event handler for the "datachannel" event on a `RTCPeerConnection` object. This event is triggered when a new data channel is created. The decorated function will be called with the data channel as an argument, allowing you to handle data communication between peers.

```python
from aiortc import RTCPeerConnection, RTCDataChannel

pc = RTCPeerConnection()

@pc.on("datachannel")
async def on_datachannel(channel: RTCDataChannel):
    print(f"📡 Data channel created: {channel.label}")
    channel.on("message", lambda msg: print(f"📥 Message received: {msg}"))
```

**`@pc.on("connectionstatechange")`**

The `@pc.on("connectionstatechange")` decorator is used to register an event handler for the "connectionstatechange" event on a `RTCPeerConnection` object. This event is triggered whenever the connection state changes (e.g., connecting, connected, disconnected). The decorated function will be called without arguments, allowing you to respond to changes in the connection state.

```python

from aiortc import RTCPeerConnection

pc = RTCPeerConnection()

@pc.on("connectionstatechange")
async def on_connectionstatechange():
    print(f"🔗 Connection state changed: {pc.connectionState}")
```

**`@pc.on("icecandidate")`**

The `@pc.on("icecandidate")` decorator is used to register an event handler for the "icecandidate" event on a `RTCPeerConnection` object. This event is triggered when a new ICE candidate is discovered. The decorated function will be called with the candidate as an argument, allowing you to handle the ICE candidates for establishing a peer-to-peer connection.

```python
from aiortc import RTCPeerConnection, RTCIceCandidate
pc = RTCPeerConnection()
@pc.on("icecandidate")
async def on_icecandidate(candidate: RTCIceCandidate):
    print(f"🧊 New ICE candidate: {candidate}"
    )
```

**`@pc.on("signalingstatechange")`**

The `@pc.on("signalingstatechange")` decorator is used to register an event handler for the "signalingstatechange" event on a `RTCPeerConnection` object. This event is triggered whenever the signaling state changes (e.g., stable, have-local-offer, have-remote-offer). The decorated function will be called without arguments, allowing you to respond to changes in the signaling state.

```python
from aiortc import RTCPeerConnection
pc = RTCPeerConnection()

@pc.on("signalingstatechange")
async def on_signalingstatechange():
    print(f"📶 Signaling state changed: {pc.signalingState}"
    )
```

**`@track.on("ended")`**

The `@track.on("ended")` decorator is used to register an event handler for the "ended" event on a `MediaStreamTrack` object. This event is triggered when the media track has ended (e.g., when the remote peer stops sending media). The decorated function will be called without arguments, allowing you to handle the end of the media stream.

```python
from aiortc import MediaStreamTrack

track = MediaStreamTrack()

@track.on("ended")
async def on_ended():
    print("🛑 Track ended"
    )
```
