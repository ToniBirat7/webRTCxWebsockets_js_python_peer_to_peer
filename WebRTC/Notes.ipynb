## **Introduction to WebRTC**

With `WebRTC`, you can add real-time communication capabilities to your application that works on top of an open standard. It supports video, voice, and generic data to be sent between peers, allowing developers to build powerful voice- and video-communication solutions.

The technology is available on all modern browsers as well as on native clients for all major platforms. The technologies behind WebRTC are implemented as an open web standard and available as regular JavaScript APIs in all major browsers.

For native clients, like Android and iOS applications, a library is available that provides the same functionality. The WebRTC project is open-source and supported by Apple, Google, Microsoft and Mozilla, amongst others. This page is maintained by the Google WebRTC team.

<hr>

`WebRTC` works on `UDP` protocol, which is faster than `TCP` but does not guarantee delivery of packets. This makes it suitable for real-time applications like video conferencing and online gaming, where low latency is more important than reliability.


## **Difference Between `Peer-to-Peer` and `Client-Server` Architecture**

### **Peer-to-Peer Architecture:**

In a peer-to-peer architecture, each participant (peer) can directly communicate with other participants without the need for a central server. Each peer acts as both a client and a server, sharing resources and data directly with one another. This architecture is often used in applications like file sharing, video conferencing, and decentralized networks.

To establish a peer-to-peer connection, signaling is typically required to exchange connection information (like IP addresses and ports) between peers. Once the connection is established, data can be transmitted directly between the peers.

Each `Peer` can have only one connection with another `Peer`. If a `Peer` wants to connect with multiple `Peers`, it needs to establish separate connections with each of them.

### **Client-Server Architecture:**

In a client-server architecture, there is a central server that manages and controls the communication between clients. Clients send requests to the server, which processes those requests and sends back responses. The server acts as a central hub for data storage, processing, and management. This architecture is commonly used in web applications, email services, and online gaming.

In a client-server model, the server is responsible for handling multiple client connections simultaneously. Clients do not communicate directly with each other; instead, they interact with the server to access resources and services.

### **Key Differences:**

1. **Communication Model:**

   - Peer-to-Peer: Direct communication between peers.
   - Client-Server: Communication through a central server.

2. **Resource Sharing:**

   - Peer-to-Peer: Resources are shared directly between peers.
   - Client-Server: Resources are managed and provided by the server.

3. **Scalability:**

   - Peer-to-Peer: Can be more scalable as each peer contributes resources.
   - Client-Server: Scalability depends on the server's capacity and resources.

4. **Reliability:**

   - Peer-to-Peer: If one peer goes offline, it can affect the connection.
   - Client-Server: The server can provide redundancy and backup to ensure reliability.

5. **Complexity:**

   - Peer-to-Peer: Can be more complex to manage connections and data consistency.
   - Client-Server: Easier to manage and control through a central server.

## **Difference Between `WebSocket` and `WebRTC`**

### **WebSocket:**

WebSocket is a communication protocol that provides full-duplex communication channels over a single TCP connection. It is designed for real-time web applications and allows for bidirectional communication between a client (usually a web browser) and a server. WebSocket is commonly used for applications like chat applications, live updates, and online gaming.

### **WebRTC:**

WebRTC (Web Real-Time Communication) is a technology that enables peer-to-peer communication directly between web browsers and mobile applications. It allows for real-time audio, video, and data sharing without the need for an intermediary server. WebRTC is commonly used for applications like video conferencing, voice calls, and file sharing.

### **Key Differences:**

1. **Communication Model:**

   - WebSocket: Client-server communication model where the client communicates with a central server.
   - WebRTC: Peer-to-peer communication model where clients can communicate directly with each other.

2. **Use Cases:**

   - WebSocket: Suitable for applications that require real-time updates from a server, such as chat applications and live notifications.
   - WebRTC: Suitable for applications that require direct peer-to-peer communication, such as video conferencing and file sharing.

3. **Media Support:**

   - WebSocket: Primarily designed for text and binary data transmission; does not natively support audio or video streaming.
   - WebRTC: Specifically designed for real-time audio and video communication, as well as data sharing.

4. **Connection Establishment:**

   - WebSocket: Requires a handshake process to establish a connection between the client and server.
   - WebRTC: Requires signaling to exchange connection information (like IP addresses and ports) between peers before establishing a direct connection.

5. **Latency:**

   - WebSocket: Generally has low latency, but it can be affected by server load and network conditions.
   - WebRTC: Designed for low-latency communication, making it suitable for real-time applications like video calls.

## **Getting Started with WebRTC**

First we've to understand `WebRTC` works on two different technologies:

**`Media Capture Devices`**

It allows us to capture audio and video from the user's device using the `getUserMedia()` API. This API prompts the user for permission to access their camera and microphone, and if granted, it provides a stream of media that can be used in our application.

**`Peer-to-Peer Connection`**

It allows us to establish a direct connection between two users using the `RTCPeerConnection` API. This API handles the signaling process, which involves exchanging information about the media streams and network configuration between the two peers.

<hr>

`WebRTC` makes it possible to build real-time communication applications that can run directly in the browser without the need to store the data on a server. This is achieved through a combination of several technologies, including:

- **Media Capture and Streams:** `WebRTC` provides APIs like `getUserMedia()` to capture audio and video from the user's device. This allows applications to access the user's camera and microphone for real-time communication.

- **Peer-to-Peer Connections:** `WebRTC` uses the `RTCPeerConnection` API to establish direct connections between users. This allows for low-latency communication without the need for a central server to relay data.

- **Signaling:** To establish a peer-to-peer connection, `WebRTC` requires a signaling mechanism to exchange connection information (like IP addresses and ports) between peers. This is typically done using a separate signaling server, which can be implemented using technologies like WebSocket or HTTP.

- **NAT Traversal:** `WebRTC` includes mechanisms for traversing Network Address Translators (NATs) and firewalls, which are commonly used in home and corporate networks. This is achieved using protocols like STUN (Session Traversal Utilities for NAT) and TURN (Traversal Using Relays around NAT).

- **Data Channels:** `WebRTC` provides a `RTCDataChannel` API that allows for the exchange of arbitrary data between peers. This can be used for applications like file sharing or real-time gaming.

- **Security:** `WebRTC` includes built-in security features, such as encryption of media streams and data channels, to ensure that communication is secure and private.

- **Cross-Platform Support:** `WebRTC` is supported by all major web browsers, including Chrome, Firefox, Safari, and Edge, as well as on mobile platforms like Android and iOS. This allows developers to build applications that can run on a wide range of devices.

- **Adaptive Bitrate Streaming:** `WebRTC` supports adaptive bitrate streaming, which allows the quality of the media stream to be adjusted dynamically based on network conditions. This helps to ensure a smooth user experience even in situations with varying bandwidth.

## **But How Will The `Peer` Know Each Other?**

The obvious answer would be with `IP` addresses. If one `Peer` knows the `IP` address of another `Peer`, they can connect directly. But, there's a problem with this approach.

**`IP` Address Problem**

Most users are behind `NAT` (Network Address Translation) routers, which means that their devices have private `IP` addresses that are not directly accessible from the internet.

This `Private` IP address is only valid within the local network i.e. your `Home Network`. We will actually need to know the `Public` IP address of the router to have a chance of connecting to a device behind it.

But, the `Device` which wants to connect to another `Device` behind a `NAT` router does not know the `Public` IP address of the router. And since we cannot connect to a `Private` IP address from the internet, we need a way to discover the `Public` IP address of the router.

### **Way To Discover `Public` IP Address of Router**

To discover the `Public` IP address of the router, we can create a `Server` on which we will first send a request from the `Device` which wants to connect to another `Device`.

Since, the `Request` is coming from the `Device` behind the `NAT` router, the `Server` will see the `Public` IP address of the router as the source of the request. The `Server` can then send this `Public` IP address back to the `Device`, which can then use it to connect to the other `Device`.

Both, the `Device` which wants to connect and the `Device` which is being connected to will send a request to the `Server` to discover their `Public` IP addresses.

These types of servers are called **`STUN` (Session Traversal Utilities for NAT) Servers**

And,

**`ICE (Interactive Connectivity Establishment)`** is a framework used in `WebRTC` to facilitate the discovery of the best path for peer-to-peer communication between two devices, especially when they are behind `NAT` (Network Address Translation) routers or firewalls.

So, both `Client A` and `Client B` will send a request to the `Server` to discover their `Public` IP addresses. Suppose, the `Server` responds with the following information:

| Client   | Private IP Address | Public IP Address | Port |
| -------- | ------------------ | ----------------- | ---- |
| Client A | 192.168.1.2        | 203.0.113.1       | 5000 |
| Client B | 192.168.1.3        | 203.0.113.1       | 5001 |

Once both the `Clients` have known their `Public` IP addresses, they now need to exchange this information so that they can connect to each other.

For that we need a **`Signaling Server`**.

## **Signaling Server**

A `Signaling Server` is a server that facilitates the exchange of signaling messages between two or more clients in a `WebRTC` application. It is responsible for establishing and managing the peer-to-peer connections between clients.

**Note:** Eariler we've mentioned that `WebRTC` does not require an intermediary server to relay data. This is true for the actual media and data transmission, but a `Signaling Server` is still needed to exchange connection information (like `IP` addresses and ports) between peers before the direct connection can be established.

Once the `Signaling Server` has facilitated the exchange of connection information between the clients, the `Clients` no longer need the `Signaling Server` to communicate with each other. They can establish a direct peer-to-peer connection using the `RTCPeerConnection` API and exchange media and data directly.

This `Signaling Server` can be implemented using various technologies, such as `WebSocket`, `HTTP`, or any other real-time communication protocol.

### **Working of Signaling Server**

Before that we need to understand about **`SDP (Session Description Protocol)`**

**`Session Description Protocol (SDP)`**

It is a format for describing the multimedia content of a communication session, such as the codecs, formats, and network information used for the session. It is used in various communication protocols, including `WebRTC`, to negotiate the parameters of a media session between two or more participants.

Basically, it contains all the information about the `Media Streams` that a `Peer` wants to send or receive, `Public IP Address`, `Port`, and other metadata.

When a `Peer` wants to establish a connection with another `Peer`, it creates an `SDP` offer that describes the media streams it wants to send and receive. This offer is then sent to the other `Peer` through the `Signaling Server`.

We can transfer this `SDP` offer using any real-time communication protocol, such as `WebSocket` or `HTTP`.

<hr>

Now,

`Client A` will create an `SDP` offer and send it to the `Signaling Server`, which will then forward it to `Client B`. `Client B` will respond with its own `SDP` answer, which will also be sent through the `Signaling Server` back to `Client A`.

Once, both `Clients` have exchanged their `SDP` offers and answers, they can use this information to establish a direct peer-to-peer connection using the `RTCPeerConnection` API.

This way `WebRTC` can establish a direct connection using:

- `Public IP Address`

- `Signaling Server`

- `Session Description Protocol (SDP)`

### **What is `SFU` (Selective Forwarding Unit)?**

An `SFU` is a type of media server used in `WebRTC` applications to facilitate multi-party video conferencing. Unlike a traditional `MCU` (Multipoint Control Unit), which mixes all incoming media streams into a single stream before sending it to participants, an `SFU` selectively forwards individual media streams to each participant based on their capabilities and preferences.

Before `SFU`, there was `Mesh Topology` for multi-party video conferencing in `WebRTC`. In a mesh topology, each participant establishes a direct peer-to-peer connection with every other participant. While this approach works well for small groups, it becomes inefficient and resource-intensive as the number of participants increases. Each participant has to send and receive multiple streams, leading to high bandwidth consumption and increased CPU usage.

**How `SFU` Works?**

In an `SFU` architecture, each participant sends their media streams to the `SFU` server. The `SFU` then forwards these streams to other participants based on their individual needs. For example, if a participant has a low bandwidth connection, the `SFU` can choose to forward only the audio stream or a lower resolution video stream to that participant.

## **More About `STUN`, `TURN`, and `ICE`**

These are protocols and techniques used in `WebRTC` to facilitate peer-to-peer communication between devices that are behind `NAT` (Network Address Translation) routers or firewalls.

As there are different types of `NAT` configurations, it can be challenging for two devices to establish a direct connection. `STUN`, `TURN`, and `ICE` help overcome these challenges by providing mechanisms for discovering public IP addresses, relaying media streams, and negotiating connection parameters.

**Type of `NATs`**

First, `NAT` is a technique used by routers to allow multiple devices on a local network to share a single public IP address. It works by translating the private IP addresses of devices on the local network to the public IP address of the router when they communicate with the internet.

There are several types of `NAT` configurations, including:

- **Full Cone NAT:** In this type of `NAT`, once a device on the local network sends a packet to an external device, any external device can send packets back to the local device using the same public IP address and port.

- **Restricted Cone NAT:** In this type of `NAT`, an external device can only send packets back to a local device if the local device has previously sent a packet to that external device.

- **Port Restricted Cone NAT:** In this type of `NAT`, an external device can only send packets back to a local device if the local device has previously sent a packet to that external device from the same port.

- **Symmetric NAT:** In this type of `NAT`, each time a device on the local network sends a packet to an external device, a unique mapping is created for that specific destination IP address and port. This means that only the external device that received the packet can send packets back to the local device using the same public IP address and port.

- **Double NAT:** This occurs when there are two `NAT` devices in the network path, such as when a local network is connected to another network that also uses `NAT`. This can complicate the process of establishing peer-to-peer connections.

- **Symmetric UDP Firewall:** This is not a type of `NAT`, but rather a firewall configuration that restricts incoming UDP packets to only those that are part of an established connection. This can also make it difficult for devices to establish peer-to-peer connections.

### **STUN (Session Traversal Utilities for NAT)**

- `STUN` is a protocol that allows a device to discover its public IP address and the type of `NAT` it is behind. It works by sending a request to a `STUN` server, which responds with the public IP address and port that the device can use to communicate with other devices on the internet.

- It is used for `Full Cone NAT`, `Restricted Cone NAT`, and `Port Restricted Cone NAT`.

- It does not work for `Symmetric NAT` and `Symmetric UDP Firewall`.

- `STUN` by default runs on `UDP` port `3478` and for `TLS` it runs on `TCP` port `5349`.

- It is cheap to maintain as it only requires minimal resources to respond to requests. We can use free public `STUN` servers like `stun.l.google.com:19302` or we can create our own `STUN` server using libraries like `coturn` in a `Docker` container.

### **Drawbacks of `WebRTC`**

Major limitations is that the connection is `Peer to Peer`, which means that one `Peer` can connect to only one `Peer` at a time. If a `Peer` wants to connect with multiple `Peers`, it needs to establish separate connections with each of them.

For example, website like `Omegle` uses `WebRTC` to connect two random users for a video chat. Here, each user is a `Peer`, and they can only connect with one other user at a time. If a user wants to connect with another random user, they need to disconnect from the current user and establish a new connection with the new user.

<hr>


But, if we want to create a video conferencing application like `Zoom` or `Google Meet`, where multiple users can join the same call the default `Peer to Peer` architecture of `WebRTC` is not sufficient.

### **Creating Multi-Peer Connection with `WebRTC`**

To create a multi-peer connection with `WebRTC`, we use `SFU (Selective Forwarding Unit)` or `MCU (Multipoint Control Unit)` server.

So, there will be a centralized `SFU` or `MCU` server that will act as a middleman between all the `Peers`. Each `Peer` will establish a connection with the `SFU` or `MCU` server, and the server will then forward the media streams to all the other `Peers` in the call.

All the `Peers` will send their media streams to the `SFU` or `MCU` server, which will create a `Single Frame` containing all the media streams and send it to all the `Peers` in the call.
