## **Bullet Notes**

- Using `async def` we can define asynchronous functions. If called, `returns` a `coroutine` object instead of executing the function immediately.

- To run the `coroutine`, we need to use `await` keyword inside another `async` function or use `asyncio.run()` method.

- Even if we use `await` inside a `async` function and run with `asyncio.run(function())`, the code inside the function runs synchronously until and unless we create `asyncio.create_task(async_function(params))`.
