It was invented by `Google` in 2012 and later adopted by the `IETF` as a standard in 2021. `QUIC` was designed to solve the limitations of `TCP` and `HTTP/2`, providing faster connection establishment, improved congestion control, and better performance for modern web applications.

**Pre `QUIC` Era**

Before, when a client or a browser wanted to have multiple requests to a server, it had to open multiple `TCP` connections. This led to issues like:

- **Head-of-Line Blocking**: If a packet was lost, all subsequent packets had to wait for the lost packet to be retransmitted, causing delays.

- **Connection Overhead**: Each `TCP` connection required a three-way handshake, leading to increased latency.

- **Multiplexing Limitations**: `HTTP/2` introduced multiplexing over a single `TCP` connection, but it still suffered from head-of-line blocking due to `TCP`'s in-order delivery requirement.

<hr>

**Post `QUIC` Era**

Many tech giants like `Google`, `Facebook`, and `Cloudflare` have adopted `QUIC` for their services, leading to significant performance improvements. For instance, `Facebook` relies completely on `QUIC` for its mobile app traffic, resulting in faster load times and reduced latency.

<hr>

`QUIC` operates over `UDP`, allowing it to implement features like multiplexing and improved congestion control without the limitations of `TCP`. It uses a combination of techniques, including:

- **0-RTT Connection Establishment**: `QUIC` can establish a connection with zero round-trip time for repeat connections, significantly reducing latency.

- **Multiplexing**: `QUIC` allows multiple streams of data to be sent over a single connection without head-of-line blocking.

- **Improved Congestion Control**: `QUIC` implements advanced congestion control algorithms to optimize data flow and reduce packet loss.

- **Built-in Encryption**: `QUIC` integrates `TLS 1.3` for secure communication, ensuring data privacy and integrity.
