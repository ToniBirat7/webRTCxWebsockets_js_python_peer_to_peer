## **HTTP**

`HTTP` (Hypertext Transfer Protocol) is a protocol used for communication between a client (such as a web browser) and a server. It is a request-response protocol, where the client sends a request to the server, and the server responds with the requested data.

It works on a stateless model, meaning that each request is independent and does not retain any information about previous requests. This makes it simple and efficient for transferring data over the internet.

<hr>

It works on top of `TCP/IP` (Transmission Control Protocol/Internet Protocol), which is the underlying protocol for communication on the internet. `HTTP` uses a client-server architecture, where the client initiates the request and the server responds with the requested data.


## **Versions of HTTP**

There are different versions of `HTTP` because the protocol has evolved over time to meet the changing needs of the internet and web applications. Each version of `HTTP` introduces new features and improvements to enhance performance, security, and functionality.

<hr>

### **HTTP/0.9**

It was the first version of `HTTP` and was introduced in 1991. It is a text-based protocol that uses a request-response model, where the client sends a request to the server, and the server responds with the requested data. `HTTP/1` supports persistent connections, which allows multiple requests and responses to be sent over a single connection.

It did not have any `Headers` or `Status Codes`. It was `Static` meaning it was just used to fetch `HTML` pages from the server.

<hr>

### **HTTP/1.0**

It was the first version of `HTTP` to be widely adopted and is still in use today. It introduced several new features, including support for `Headers`, `Status Codes`, and `Content Types`. `HTTP/1.0` also introduced the concept of `Persistent Connections`, which allows multiple requests and responses to be sent over a single connection.

It included `GET`, `POST`, and `HEAD` methods.

**Working of HTTP/1.0**

- First there's a `TCP` `Three-Way Handshake` between the client and server to establish a connection.

- Then for `HTTPs` there is a `TLS Handshake` to establish a secure connection.

- Then the client sends an `HTTP` request to the server, which includes a `Request Line`, `Headers`, and an optional `Body`.

- The server processes the request and sends back an `HTTP` response, which includes a `Status Line`, `Headers`, and an optional `Body`. And then the server closes the connection.

**Problems with HTTP/1.0**

But, when a client makes a request to the server, the server closes the connection after sending the response. This means that for each new request, a new connection must be established, which can lead to increased latency and overhead.

<hr>

### **HTTP/1.1**

To fix the problems of `HTTP/1.0`, `HTTP/1.1` was introduced in 1997. It introduced several new features, such as `Persistent Connections` by default, `Chunked Transfer Encoding`, and `Pipelining`. `HTTP/1.1` also introduced new `Headers`, such as `Host`, which allows multiple domains to be hosted on a single server.

It also introduced features like `Caching`, `Content Negotiation`, and `Range Requests`.

**Persistent Connections**

In `HTTP/1.1`, persistent connections are enabled by default, which means that the connection between the client and server remains open after a request is completed. This allows multiple requests and responses to be sent over a single connection, reducing the overhead of establishing new connections for each request.

**Pipelining**

Pipelining allows multiple requests to be sent over a single connection without waiting for the response to each request. This can improve performance by reducing the latency between requests and responses.

However, pipelining is not widely supported by all HTTP/1.1 clients and servers, and it can lead to `Head-of-Line` blocking, where a slow response to one request delays all subsequent responses.

**Chunked Transfer Encoding**

Chunked transfer encoding allows the server to send a response in chunks, rather than sending the entire response at once. This can be useful for large responses or for streaming data, as it allows the client to start processing the response before the entire response has been received.

**`Problems with HTTP/1.1`**

- `Head-of-Line Blocking`: In `HTTP/1.1`, if a request is delayed or takes a long time to process, it can block subsequent requests from being processed, leading to increased latency and reduced performance.

- `Limited Multiplexing`: While `HTTP/1.1` supports persistent connections and pipelining, it still has limitations when it comes to multiplexing multiple requests and responses over a single connection. This can lead to inefficiencies and increased latency in certain scenarios.

- `Full Duplex Communication`: `HTTP/1.1` is primarily designed for request-response communication, which means that the client sends a request and waits for a response from the server. This can be limiting in scenarios where real-time communication is required, such as in chat applications or online gaming.

### **HTTP/2**

To address the limitations of `HTTP/1.1`, `HTTP/2` was introduced in 2015. It is a major revision of the `HTTP` protocol and introduces several new features, such as `Binary Framing`, `Multiplexing`, and `Header Compression`. `HTTP/2` also introduces new `Headers`, such as `:method`, `:path`, and `:authority`, which are used to convey information about the request.

**Binary Framing**

Unlike `HTTP/1.1`, which is a text-based protocol, `HTTP/2` uses a binary framing layer to encode and transmit data. This allows for more efficient parsing and processing of requests and responses, as well as improved performance.

Messages are broken down into smaller frames, which can be interleaved and sent over a single connection. This allows multiple requests and responses to be sent simultaneously, improving performance and reducing latency.

**Multiplexing**

Multiplexing allows multiple requests and responses to be sent over a single connection simultaneously. This is achieved by breaking down messages into smaller frames, which can be interleaved and sent over the connection. This allows for more efficient use of network resources and reduces latency.

This allowed for protocols like `WebSocket` to be implemented over `HTTP/2`.

This also fixed the `Head-of-Line Blocking` problem of `HTTP/1.1`.

**Stream Prioritization**

`HTTP/2` allows clients to prioritize streams, which means that certain requests can be given higher priority than others. This can be useful in scenarios where certain requests are more important than others, such as in video streaming or online gaming.

**Header Compression**

`HTTP/2` uses a binary encoding format for headers, which allows for more efficient compression and transmission of header data. This can reduce the amount of data that needs to be transmitted over the network, improving performance and reducing latency.

It uses `HPACK` compression algorithm to compress headers.

**`Problems with HTTP/2`**

`TCPs` nature is handling packet loss by retransmitting lost packets and `Head-of-Line Blocking` can still occur at the `TCP` layer. If a packet is lost, all subsequent packets must wait for the lost packet to be retransmitted before they can be processed, which can lead to increased latency and reduced performance.

To address the limitations of `HTTP/2`, `HTTP/3` was introduced in 2022. It is a major revision of the `HTTP` protocol and introduces several new features, such as `QUIC` transport protocol, improved security, and better performance.

`HTTP/3` uses `QUIC` (Quick UDP Internet Connections) as its transport protocol instead of `TCP`, which is designed to be faster and more efficient than `TCP`. `QUIC` uses `UDP` (User Datagram Protocol) as its underlying transport protocol, which allows for faster connection establishment and improved performance in scenarios with high latency or packet loss.

`QUIC` was developed by `Google` and is designed to provide a more efficient and reliable transport protocol for web applications.

`QUIC` is based on `UDP`, which is a connectionless protocol which does not needs to establish a connection before transmitting data. This allows for faster connection establishment and reduced latency.

`QUIC` provides several benefits over `TCP`, including:

- Faster connection establishment: `QUIC` uses a `0-RTT` (Zero Round Trip Time) connection establishment process, which allows for faster connection setup and reduced latency.

- Improved performance in high-latency or lossy networks: `QUIC` is designed to perform better in scenarios with high latency or packet loss, which can improve the performance of web applications in these environments.

- Built-in encryption: `QUIC` includes built-in encryption using `TLS 1.3`, which provides improved security and privacy for web applications.

- Network Chnage Resilience: `QUIC` is designed to handle network changes, such as switching between Wi-Fi and cellular networks, without dropping the connection. This can improve the user experience for mobile applications.

## **Protocols for Real-Time Communication**

`Real Time Communication` refers to the exchange of information between two or more parties in a way that allows for immediate interaction and response. This type of communication is often used in applications such as video conferencing, online gaming, and instant messaging.

We cannot use `HTTP` for real-time communication because it is a request-response protocol, which means that the client must send a request to the server and wait for a response before it can continue. This can introduce latency and delays in communication, which is not ideal for real-time applications.

To enable real-time communication, we can use protocols such as `WebSocket`, `WebRTC`, and `QUIC`. These protocols allow for bidirectional communication between the client and server, which means that both parties can send and receive data at the same time without the need for a request-response cycle.

<hr>

There are two types of `Protocols` which works in the `Transport Layer` of the `OSI Model`:

### **TCP (Transmission Control Protocol)**

`TCP` is a connection-oriented protocol that provides reliable, ordered, and error-checked delivery of data between applications running on hosts communicating via an `IP` network. It establishes a connection between the sender and receiver before transmitting data, ensuring that all packets are received in the correct order and without errors. If any packets are lost or corrupted during transmission, `TCP` will retransmit them until they are successfully received.

**Working of `TCP`:**

1. **Connection Establishment:** `TCP` uses a three-way handshake to establish a connection between the sender and receiver. The sender sends a `SYN` (synchronize) packet to the receiver, which responds with a `SYN-ACK` (synchronize-acknowledge) packet. The sender then sends an `ACK` (acknowledge) packet to complete the handshake.

2. **Data Transmission:** Once the connection is established, data can be transmitted between the sender and receiver. `TCP` breaks the data into smaller packets and assigns a sequence number to each packet to ensure that they are received in the correct order. The receiver sends an acknowledgment for each packet received, and if any packets are lost or corrupted during transmission, `TCP` will retransmit them until they are successfully received.

3. **Connection Termination:** When the data transmission is complete, either the sender or receiver can initiate the connection termination process. This involves a four-way handshake, where the sender sends a `FIN` (finish) packet to the receiver, which responds with an `ACK` packet. The receiver then sends a `FIN` packet to the sender, which responds with an `ACK` packet to complete the termination process.

### **UDP (User Datagram Protocol)**

`UDP` is a connectionless protocol that provides a simple and efficient way to transmit data between applications running on hosts communicating via an `IP` network. Unlike `TCP`, `UDP` does not establish a connection between the sender and receiver before transmitting data, and it does not guarantee the delivery or order of packets.

**Working of `UDP`:**

1. **Data Transmission:** `UDP` transmits data in the form of datagrams, which are small packets of data that contain the source and destination addresses, as well as the data itself. Each datagram is independent and can be sent in any order, without the need for a connection or acknowledgment from the receiver.

2. **No Connection Establishment:** Since `UDP` is a connectionless protocol, there is no need for a handshake or connection establishment process before transmitting data. The sender can simply send datagrams to the receiver without any prior communication.

3. **No Acknowledgment or Retransmission:** `UDP` does not provide any mechanism for acknowledging the receipt of datagrams or retransmitting lost or corrupted packets. This means that if a datagram is lost or corrupted during transmission, it will not be retransmitted, and the receiver will not be notified of the loss.
