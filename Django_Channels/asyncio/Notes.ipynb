## **Bullet Notes**

- Using `async def` we can define asynchronous functions. If called, `returns` a `coroutine` object instead of executing the function immediately.

- To run the `coroutine`, we need to use `await` keyword inside another `async` function or use `asyncio.run()` method.

- Even if we use `await` inside a `async` function and run with `asyncio.run(function())`, the code inside the function runs synchronously until and unless we create `asyncio.create_task(async_function(params))`.

- `asyncio.sleep(seconds)` is a non-blocking sleep function that allows other tasks to run while waiting.

- `asyncio.gather(*tasks)` is used to run multiple asynchronous tasks concurrently and wait for all of them to complete. We can collect the results in a `Single variable` then `iterate` over it to get individual results.

- We use `async with asyncio.TaskGroup() as tg:` to create a task group that manages multiple asynchronous tasks. Tasks added to the group will be automatically awaited when the group is exited.

**`Future`**

- A `Future` is a low-level construct that represents a value that may not be available yet. It is used to bridge the gap between callback-based code and `async`/`await` syntax.

- We can create a `Future` object using `asyncio.Future()`. We can set its result using `future.set_result(value)` and retrieve the result using `await future`.

- `Future` objects are typically used in more advanced scenarios where you need fine-grained control over the completion of an asynchronous operation.

**`Synchronization Primitives`**

- `asyncio` provides several synchronization primitives to manage access to shared resources in an asynchronous environment. Some commonly used primitives include:

  - `Lock`: A lock that can be acquired and released by asynchronous tasks to ensure exclusive access to a shared resource.

  - `Event`: An event that can be set or cleared by one task and waited for by other tasks.

  - `Semaphore`: A semaphore that limits the number of concurrent accesses to a shared resource.

  - `Condition`: A condition variable that allows tasks to wait for certain conditions to be met before proceeding.
