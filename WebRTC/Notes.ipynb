## **Introduction to WebRTC**

With `WebRTC`, you can add real-time communication capabilities to your application that works on top of an open standard. It supports video, voice, and generic data to be sent between peers, allowing developers to build powerful voice- and video-communication solutions.

The technology is available on all modern browsers as well as on native clients for all major platforms. The technologies behind WebRTC are implemented as an open web standard and available as regular JavaScript APIs in all major browsers.

For native clients, like Android and iOS applications, a library is available that provides the same functionality. The WebRTC project is open-source and supported by Apple, Google, Microsoft and Mozilla, amongst others. This page is maintained by the Google WebRTC team.

<hr>

`WebRTC` works on `UDP` protocol, which is faster than `TCP` but does not guarantee delivery of packets. This makes it suitable for real-time applications like video conferencing and online gaming, where low latency is more important than reliability.


## **Difference Between `Peer-to-Peer` and `Client-Server` Architecture**

### **Peer-to-Peer Architecture:**

In a peer-to-peer architecture, each participant (peer) can directly communicate with other participants without the need for a central server. Each peer acts as both a client and a server, sharing resources and data directly with one another. This architecture is often used in applications like file sharing, video conferencing, and decentralized networks.

To establish a peer-to-peer connection, signaling is typically required to exchange connection information (like IP addresses and ports) between peers. Once the connection is established, data can be transmitted directly between the peers.

Each `Peer` can have only one connection with another `Peer`. If a `Peer` wants to connect with multiple `Peers`, it needs to establish separate connections with each of them.

### **Client-Server Architecture:**

In a client-server architecture, there is a central server that manages and controls the communication between clients. Clients send requests to the server, which processes those requests and sends back responses. The server acts as a central hub for data storage, processing, and management. This architecture is commonly used in web applications, email services, and online gaming.

In a client-server model, the server is responsible for handling multiple client connections simultaneously. Clients do not communicate directly with each other; instead, they interact with the server to access resources and services.

### **Key Differences:**

1. **Communication Model:**

   - Peer-to-Peer: Direct communication between peers.
   - Client-Server: Communication through a central server.

2. **Resource Sharing:**

   - Peer-to-Peer: Resources are shared directly between peers.
   - Client-Server: Resources are managed and provided by the server.

3. **Scalability:**

   - Peer-to-Peer: Can be more scalable as each peer contributes resources.
   - Client-Server: Scalability depends on the server's capacity and resources.

4. **Reliability:**

   - Peer-to-Peer: If one peer goes offline, it can affect the connection.
   - Client-Server: The server can provide redundancy and backup to ensure reliability.

5. **Complexity:**

   - Peer-to-Peer: Can be more complex to manage connections and data consistency.
   - Client-Server: Easier to manage and control through a central server.

## **Difference Between `WebSocket` and `WebRTC`**

### **WebSocket:**

WebSocket is a communication protocol that provides full-duplex communication channels over a single TCP connection. It is designed for real-time web applications and allows for bidirectional communication between a client (usually a web browser) and a server. WebSocket is commonly used for applications like chat applications, live updates, and online gaming.

### **WebRTC:**

WebRTC (Web Real-Time Communication) is a technology that enables peer-to-peer communication directly between web browsers and mobile applications. It allows for real-time audio, video, and data sharing without the need for an intermediary server. WebRTC is commonly used for applications like video conferencing, voice calls, and file sharing.

### **Key Differences:**

1. **Communication Model:**

   - WebSocket: Client-server communication model where the client communicates with a central server.
   - WebRTC: Peer-to-peer communication model where clients can communicate directly with each other.

2. **Use Cases:**

   - WebSocket: Suitable for applications that require real-time updates from a server, such as chat applications and live notifications.
   - WebRTC: Suitable for applications that require direct peer-to-peer communication, such as video conferencing and file sharing.

3. **Media Support:**

   - WebSocket: Primarily designed for text and binary data transmission; does not natively support audio or video streaming.
   - WebRTC: Specifically designed for real-time audio and video communication, as well as data sharing.

4. **Connection Establishment:**

   - WebSocket: Requires a handshake process to establish a connection between the client and server.
   - WebRTC: Requires signaling to exchange connection information (like IP addresses and ports) between peers before establishing a direct connection.

5. **Latency:**

   - WebSocket: Generally has low latency, but it can be affected by server load and network conditions.
   - WebRTC: Designed for low-latency communication, making it suitable for real-time applications like video calls.

## **Getting Started with WebRTC**

First we've to understand `WebRTC` works on two different technologies:

**`Media Capture Devices`**

It allows us to capture audio and video from the user's device using the `getUserMedia()` API. This API prompts the user for permission to access their camera and microphone, and if granted, it provides a stream of media that can be used in our application.

**`Peer-to-Peer Connection`**

It allows us to establish a direct connection between two users using the `RTCPeerConnection` API. This API handles the signaling process, which involves exchanging information about the media streams and network configuration between the two peers.
